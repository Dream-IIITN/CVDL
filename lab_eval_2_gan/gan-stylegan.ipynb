```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from PIL import Image
import cv2
from skimage.metrics import structural_similarity as ssim
from tensorflow.keras.applications.inception_v3 import InceptionV3
from scipy.linalg import sqrtm
import time

# Set random seed for reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# Configuration
IMAGE_SIZE = 64  # Using 64x64 for faster training (original was 2304x4096)
BATCH_SIZE = 32
EPOCHS = 250  # Adjust based on your time constraints
LATENT_DIM = 100
NUM_CLASSES = 7
NUM_SAMPLES_PER_CLASS = 15  # ~100 total (15*7=105)
CHECKPOINT_DIR = './training_checkpoints'
OUTPUT_DIR = './generated_images'

# Create directories if they don't exist
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Class names (from your directory structure)
class_names = [
    'ANTHRACNOSE',
    'BACTERIAL CRANKER',
    'DIEBACK',
    'GALL MILDGE',
    'HEALTHY',
    'MANGO SOOTY',
    'SOOTY MOULD'
]

# Data Loading and Preprocessing
def load_and_preprocess_dataset(data_dir):
    images = []
    labels = []
    
    try:
        for class_idx, class_name in enumerate(class_names):
            class_dir = os.path.join(data_dir, class_name)
            if not os.path.exists(class_dir):
                print(f"Warning: Directory {class_dir} not found. Skipping.")
                continue
                
            for img_name in os.listdir(class_dir):
                img_path = os.path.join(class_dir, img_name)
                try:
                    # Load and resize image
                    img = Image.open(img_path).convert('RGB')
                    img = img.resize((IMAGE_SIZE, IMAGE_SIZE))
                    img_array = np.array(img)
                    
                    # Normalize to [-1, 1] range for GAN training
                    img_array = (img_array.astype(np.float32) - 127.5) / 127.5
                    
                    images.append(img_array)
                    labels.append(class_idx)
                except Exception as e:
                    print(f"Error processing {img_path}: {str(e)}")
                    continue
        
        return np.array(images), np.array(labels)
    except Exception as e:
        print(f"Error in dataset loading: {str(e)}")
        return None, None

# Generator Model
def make_generator_model():
    model = tf.keras.Sequential()
    model.add(layers.Dense(8*8*256, use_bias=False, input_shape=(LATENT_DIM,)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    
    model.add(layers.Reshape((8, 8, 256)))
    
    model.add(layers.Conv2DTranspose(128, (5, 5), strides=(1, 1), padding='same', use_bias=False))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    
    model.add(layers.Conv2DTranspose(64, (5, 5), strides=(2, 2), padding='same', use_bias=False))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    
    model.add(layers.Conv2DTranspose(3, (5, 5), strides=(2, 2), padding='same', use_bias=False, activation='tanh'))
    
    return model

def make_discriminator_model():
    model = tf.keras.Sequential()
    model.add(layers.Conv2D(64, (5, 5), strides=(2, 2), padding='same', 
                            input_shape=[IMAGE_SIZE, IMAGE_SIZE, 3]))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    
    model.add(layers.Conv2D(128, (5, 5), strides=(2, 2), padding='same'))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    
    model.add(layers.Conv2D(256, (5, 5), strides=(2, 2), padding='same'))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    
    # This is the critical change - add GlobalAveragePooling2D instead of Flatten
    model.add(layers.GlobalAveragePooling2D())
    model.add(layers.Dense(1))
    
    return model
# Loss functions
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

def discriminator_loss(real_output, fake_output):
    real_loss = cross_entropy(tf.ones_like(real_output), real_output)
    fake_loss = cross_entropy(tf.zeros_like(fake_output), fake_output)
    total_loss = real_loss + fake_loss
    return total_loss

def generator_loss(fake_output):
    return cross_entropy(tf.ones_like(fake_output), fake_output)

# Training
@tf.function
def train_step(images, generator, discriminator, generator_optimizer, discriminator_optimizer):
    noise = tf.random.normal([BATCH_SIZE, LATENT_DIM])
    
    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator(noise, training=True)
        
        real_output = discriminator(images, training=True)
        fake_output = discriminator(generated_images, training=True)
        
        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)
    
    gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
    gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)
    
    generator_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
    discriminator_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))
    
    return gen_loss, disc_loss

def train(dataset, epochs, generator, discriminator, 
          generator_optimizer, discriminator_optimizer): 
    # Initialize models to avoid layout errors
    test_noise = tf.random.normal([1, LATENT_DIM])
    _ = generator(test_noise)
    _ = discriminator(tf.random.normal([1, IMAGE_SIZE, IMAGE_SIZE, 3]))
    
    for epoch in range(epochs):
        start = time.time()
        gen_loss_list = []
        disc_loss_list = []
        
        for image_batch in dataset:
            gen_loss, disc_loss = train_step(image_batch, generator, discriminator,
                                           generator_optimizer, discriminator_optimizer)
            gen_loss_list.append(gen_loss)
            disc_loss_list.append(disc_loss)
        
        # Print progress every epoch
        print(f'Epoch {epoch + 1}, '
              f'Gen Loss: {np.mean(gen_loss_list):.4f}, '
              f'Disc Loss: {np.mean(disc_loss_list):.4f}, '
              f'Time: {time.time() - start:.2f}s')
        
        # Generate sample images every 5 epochs
        if (epoch + 1) % 5 == 0:
            test_noise = tf.random.normal([NUM_SAMPLES_PER_CLASS, LATENT_DIM])
            generated_images = generator(test_noise, training=False)
            
            # Display images
            plt.figure(figsize=(10, 10))
            for i in range(generated_images.shape[0]):
                plt.subplot(5, 3, i+1)
                plt.imshow((generated_images[i] * 0.5 + 0.5))
                plt.axis('off')
            plt.show()
# Generate and save images
def generate_and_save_images(model, epoch, test_input):
    predictions = model(test_input, training=False)
    
    fig = plt.figure(figsize=(10, 10))
    
    for i in range(predictions.shape[0]):
        plt.subplot(5, 3, i+1)
        plt.imshow((predictions[i] * 0.5 + 0.5))  # Scale from [-1, 1] to [0, 1]
        plt.axis('off')
    
    plt.savefig(os.path.join(OUTPUT_DIR, f'image_at_epoch_{epoch:04d}.png'))
    plt.close()

# Evaluation Metrics
def calculate_ssim(real_images, generated_images):
    ssim_values = []
    min_size = min(len(real_images), len(generated_images))
    
    for i in range(min_size):
        try:
            # Convert to grayscale
            real_gray = cv2.cvtColor(real_images[i], cv2.COLOR_RGB2GRAY)
            gen_gray = cv2.cvtColor(generated_images[i], cv2.COLOR_RGB2GRAY)
            
            # Ensure identical dimensions
            h, w = real_gray.shape
            gen_gray = cv2.resize(gen_gray, (w, h))
            
            # Calculate SSIM with proper data range
            data_range = 255 if real_gray.dtype == np.uint8 else 2.0
            ssim_val = ssim(real_gray, gen_gray, 
                          data_range=data_range)
            ssim_values.append(ssim_val)
        except Exception as e:
            print(f"Error calculating SSIM for image {i}: {str(e)}")
            continue
    
    return np.mean(ssim_values) if ssim_values else 0.0

def calculate_fid(real_images, generated_images):
    # Load InceptionV3 model
    inception = InceptionV3(include_top=False, pooling='avg', input_shape=(299, 299, 3))
    
    # Resize images to 299x299 for InceptionV3
    real_resized = [cv2.resize(img, (299, 299)) for img in real_images]
    gen_resized = [cv2.resize(img, (299, 299)) for img in generated_images]
    
    # Convert to numpy arrays
    real_array = np.array(real_resized)
    gen_array = np.array(gen_resized)
    
    # Scale pixel values
    real_array = (real_array + 1) * 127.5  # [-1,1] to [0,255]
    gen_array = (gen_array + 1) * 127.5
    
    # Get activations
    act_real = inception.predict(real_array)
    act_gen = inception.predict(gen_array)
    
    # Calculate mean and covariance
    mu_real, sigma_real = act_real.mean(axis=0), np.cov(act_real, rowvar=False)
    mu_gen, sigma_gen = act_gen.mean(axis=0), np.cov(act_gen, rowvar=False)
    
    # Calculate FID
    ssdiff = np.sum((mu_real - mu_gen)**2.0)
    covmean = sqrtm(sigma_real.dot(sigma_gen))
    
    if np.iscomplexobj(covmean):
        covmean = covmean.real
    
    fid = ssdiff + np.trace(sigma_real + sigma_gen - 2.0 * covmean)
    return fid

# Main Execution
def main():
    # Load dataset
    data_dir = '/kaggle/input/mangoleaf-dataset/dataset'  # Change this to your dataset path
    print("Loading dataset...")
    images, labels = load_and_preprocess_dataset(data_dir)
    
    if images is None or len(images) == 0:
        print("Error: No images loaded. Check your dataset path.")
        return
    
    print(f"Loaded {len(images)} images")
    
    # Create TensorFlow dataset
    dataset = tf.data.Dataset.from_tensor_slices(images).shuffle(len(images)).batch(BATCH_SIZE)
    
    # Create models
    generator = make_generator_model()
    discriminator = make_discriminator_model()
    
    # Optimizers
    generator_optimizer = tf.keras.optimizers.Adam(1e-4)
    discriminator_optimizer = tf.keras.optimizers.Adam(1e-4)
    
    # Checkpoint setup
    checkpoint_prefix = os.path.join(CHECKPOINT_DIR, "ckpt")
    checkpoint = tf.train.Checkpoint(generator_optimizer=generator_optimizer,
                                   discriminator_optimizer=discriminator_optimizer,
                                   generator=generator,
                                   discriminator=discriminator)
    
    # Train the model
    print("Starting training...")
    try:
        train(dataset, EPOCHS, generator, discriminator,
      generator_optimizer, discriminator_optimizer)
    except Exception as e:
        print(f"Error during training: {str(e)}")
        return
    
    # Generate final synthetic images
    print("Generating synthetic images...")
    noise = tf.random.normal([NUM_SAMPLES_PER_CLASS * NUM_CLASSES, LATENT_DIM])
    generated_images = generator(noise, training=False).numpy()
    
    # Save generated images
    os.makedirs(os.path.join(OUTPUT_DIR, 'final'), exist_ok=True)
    for i, img in enumerate(generated_images):
        img = (img * 127.5 + 127.5).astype(np.uint8)  # Convert to 0-255 range
        img_pil = Image.fromarray(img)
        class_idx = i // NUM_SAMPLES_PER_CLASS
        class_name = class_names[class_idx]
        img_pil.save(os.path.join(OUTPUT_DIR, 'final', f'{class_name}_{i}.png'))
    
    print(f"Generated {len(generated_images)} images saved to {OUTPUT_DIR}/final")
    
    # Evaluation (using a subset of real images)
    print("Calculating evaluation metrics...")
    real_subset = images[:100]  # Use first 100 real images for comparison
    gen_subset = generated_images[:100]
    
    # Calculate SSIM
    ssim_score = calculate_ssim(real_subset, gen_subset)
    print(f"SSIM Score: {ssim_score:.4f}")
    
    # Calculate FID (may take some time)
    fid_score = calculate_fid(real_subset, gen_subset)
    print(f"FID Score: {fid_score:.2f}")
    
    # Save metrics to file
    with open(os.path.join(OUTPUT_DIR, 'metrics.txt'), 'w') as f:
        f.write(f"SSIM: {ssim_score:.4f}\n")
        f.write(f"FID: {fid_score:.2f}\n")
    
    print("All done!")

if __name__ == "__main__":
    main()
```

```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from PIL import Image
import cv2
from skimage.metrics import structural_similarity as ssim
from tensorflow.keras.applications.inception_v3 import InceptionV3
from scipy.linalg import sqrtm
import time

# Set random seed for reproducibility
tf.random.set_seed(42)
np.random.seed(42)

# Configuration
IMAGE_SIZE = 64  # Using 64x64 for faster training (original was 2304x4096)
BATCH_SIZE = 32
EPOCHS = 10  # Adjust based on your time constraints
LATENT_DIM = 100
NUM_CLASSES = 7
NUM_SAMPLES_PER_CLASS = 15  # ~100 total (15*7=105)
CHECKPOINT_DIR = './training_checkpoints'
OUTPUT_DIR = './generated_images'

# Create directories if they don't exist
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Class names (from your directory structure)
class_names = [
    'ANTHRACNOSE',
    'BACTERIAL CRANKER',
    'DIEBACK',
    'GALL MILDGE',
    'HEALTHY',
    'MANGO SOOTY',
    'SOOTY MOULD'
]

# Data Loading and Preprocessing
def load_and_preprocess_dataset(data_dir):
    images = []
    labels = []
    
    try:
        for class_idx, class_name in enumerate(class_names):
            class_dir = os.path.join(data_dir, class_name)
            if not os.path.exists(class_dir):
                print(f"Warning: Directory {class_dir} not found. Skipping.")
                continue
                
            for img_name in os.listdir(class_dir):
                img_path = os.path.join(class_dir, img_name)
                try:
                    # Load and resize image
                    img = Image.open(img_path).convert('RGB')
                    img = img.resize((IMAGE_SIZE, IMAGE_SIZE))
                    img_array = np.array(img)
                    
                    # Normalize to [-1, 1] range for GAN training
                    img_array = (img_array.astype(np.float32) - 127.5) / 127.5
                    
                    images.append(img_array)
                    labels.append(class_idx)
                except Exception as e:
                    print(f"Error processing {img_path}: {str(e)}")
                    continue
        
        return np.array(images), np.array(labels)
    except Exception as e:
        print(f"Error in dataset loading: {str(e)}")
        return None, None

# Generator Model
def make_generator_model():
    model = tf.keras.Sequential()
    model.add(layers.Dense(8*8*256, use_bias=False, input_shape=(LATENT_DIM,)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    
    model.add(layers.Reshape((8, 8, 256)))
    
    model.add(layers.Conv2DTranspose(128, (5, 5), strides=(1, 1), padding='same', use_bias=False))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    
    model.add(layers.Conv2DTranspose(64, (5, 5), strides=(2, 2), padding='same', use_bias=False))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    
    model.add(layers.Conv2DTranspose(3, (5, 5), strides=(2, 2), padding='same', use_bias=False, activation='tanh'))
    
    return model

def make_discriminator_model():
    model = tf.keras.Sequential()
    model.add(layers.Conv2D(64, (5, 5), strides=(2, 2), padding='same', 
                            input_shape=[IMAGE_SIZE, IMAGE_SIZE, 3]))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    
    model.add(layers.Conv2D(128, (5, 5), strides=(2, 2), padding='same'))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    
    model.add(layers.Conv2D(256, (5, 5), strides=(2, 2), padding='same'))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    
    # This is the critical change - add GlobalAveragePooling2D instead of Flatten
    model.add(layers.GlobalAveragePooling2D())
    model.add(layers.Dense(1))
    
    return model
# Loss functions
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

def discriminator_loss(real_output, fake_output):
    real_loss = cross_entropy(tf.ones_like(real_output), real_output)
    fake_loss = cross_entropy(tf.zeros_like(fake_output), fake_output)
    total_loss = real_loss + fake_loss
    return total_loss

def generator_loss(fake_output):
    return cross_entropy(tf.ones_like(fake_output), fake_output)

# Training
@tf.function
def train_step(images, generator, discriminator, generator_optimizer, discriminator_optimizer):
    noise = tf.random.normal([BATCH_SIZE, LATENT_DIM])
    
    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator(noise, training=True)
        
        real_output = discriminator(images, training=True)
        fake_output = discriminator(generated_images, training=True)
        
        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)
    
    gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
    gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)
    
    generator_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
    discriminator_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))
    
    return gen_loss, disc_loss

def train(dataset, epochs, generator, discriminator, 
          generator_optimizer, discriminator_optimizer): 
    # Initialize models to avoid layout errors
    test_noise = tf.random.normal([1, LATENT_DIM])
    _ = generator(test_noise)
    _ = discriminator(tf.random.normal([1, IMAGE_SIZE, IMAGE_SIZE, 3]))
    
    for epoch in range(epochs):
        start = time.time()
        gen_loss_list = []
        disc_loss_list = []
        
        for image_batch in dataset:
            gen_loss, disc_loss = train_step(image_batch, generator, discriminator,
                                           generator_optimizer, discriminator_optimizer)
            gen_loss_list.append(gen_loss)
            disc_loss_list.append(disc_loss)
        
        # Print progress every epoch
        print(f'Epoch {epoch + 1}, '
              f'Gen Loss: {np.mean(gen_loss_list):.4f}, '
              f'Disc Loss: {np.mean(disc_loss_list):.4f}, '
              f'Time: {time.time() - start:.2f}s')
        
        # Generate sample images every 5 epochs
        if (epoch + 1) % 5 == 0:
            test_noise = tf.random.normal([NUM_SAMPLES_PER_CLASS, LATENT_DIM])
            generated_images = generator(test_noise, training=False)
            
            # Display images
            plt.figure(figsize=(10, 10))
            for i in range(generated_images.shape[0]):
                plt.subplot(5, 3, i+1)
                plt.imshow((generated_images[i] * 0.5 + 0.5))
                plt.axis('off')
            plt.show()
# Generate and save images
def generate_and_save_images(model, epoch, test_input):
    predictions = model(test_input, training=False)
    
    fig = plt.figure(figsize=(10, 10))
    
    for i in range(predictions.shape[0]):
        plt.subplot(5, 3, i+1)
        plt.imshow((predictions[i] * 0.5 + 0.5))  # Scale from [-1, 1] to [0, 1]
        plt.axis('off')
    
    plt.savefig(os.path.join(OUTPUT_DIR, f'image_at_epoch_{epoch:04d}.png'))
    plt.close()

# Evaluation Metrics
def calculate_ssim(real_images, generated_images):
    ssim_values = []
    min_size = min(len(real_images), len(generated_images))
    
    for i in range(min_size):
        try:
            # Convert to grayscale
            real_gray = cv2.cvtColor(real_images[i], cv2.COLOR_RGB2GRAY)
            gen_gray = cv2.cvtColor(generated_images[i], cv2.COLOR_RGB2GRAY)
            
            # Ensure identical dimensions
            h, w = real_gray.shape
            gen_gray = cv2.resize(gen_gray, (w, h))
            
            # Calculate SSIM with proper data range
            data_range = 255 if real_gray.dtype == np.uint8 else 2.0
            ssim_val = ssim(real_gray, gen_gray, 
                          data_range=data_range)
            ssim_values.append(ssim_val)
        except Exception as e:
            print(f"Error calculating SSIM for image {i}: {str(e)}")
            continue
    
    return np.mean(ssim_values) if ssim_values else 0.0

def calculate_fid(real_images, generated_images):
    # Load InceptionV3 model
    inception = InceptionV3(include_top=False, pooling='avg', input_shape=(299, 299, 3))
    
    # Resize images to 299x299 for InceptionV3
    real_resized = [cv2.resize(img, (299, 299)) for img in real_images]
    gen_resized = [cv2.resize(img, (299, 299)) for img in generated_images]
    
    # Convert to numpy arrays
    real_array = np.array(real_resized)
    gen_array = np.array(gen_resized)
    
    # Scale pixel values
    real_array = (real_array + 1) * 127.5  # [-1,1] to [0,255]
    gen_array = (gen_array + 1) * 127.5
    
    # Get activations
    act_real = inception.predict(real_array)
    act_gen = inception.predict(gen_array)
    
    # Calculate mean and covariance
    mu_real, sigma_real = act_real.mean(axis=0), np.cov(act_real, rowvar=False)
    mu_gen, sigma_gen = act_gen.mean(axis=0), np.cov(act_gen, rowvar=False)
    
    # Calculate FID
    ssdiff = np.sum((mu_real - mu_gen)**2.0)
    covmean = sqrtm(sigma_real.dot(sigma_gen))
    
    if np.iscomplexobj(covmean):
        covmean = covmean.real
    
    fid = ssdiff + np.trace(sigma_real + sigma_gen - 2.0 * covmean)
    return fid

# Main Execution
def main():
    # Load dataset
    data_dir = '/kaggle/input/mangoleaf-dataset/dataset'  # Change this to your dataset path
    print("Loading dataset...")
    images, labels = load_and_preprocess_dataset(data_dir)
    
    if images is None or len(images) == 0:
        print("Error: No images loaded. Check your dataset path.")
        return
    
    print(f"Loaded {len(images)} images")
    
    # Create TensorFlow dataset
    dataset = tf.data.Dataset.from_tensor_slices(images).shuffle(len(images)).batch(BATCH_SIZE)
    
    # Create models
    generator = make_generator_model()
    discriminator = make_discriminator_model()
    
    # Optimizers
    generator_optimizer = tf.keras.optimizers.Adam(1e-4)
    discriminator_optimizer = tf.keras.optimizers.Adam(1e-4)
    
    # Checkpoint setup
    checkpoint_prefix = os.path.join(CHECKPOINT_DIR, "ckpt")
    checkpoint = tf.train.Checkpoint(generator_optimizer=generator_optimizer,
                                   discriminator_optimizer=discriminator_optimizer,
                                   generator=generator,
                                   discriminator=discriminator)
    
    # Train the model
    print("Starting training...")
    try:
        train(dataset, EPOCHS, generator, discriminator,
      generator_optimizer, discriminator_optimizer)
    except Exception as e:
        print(f"Error during training: {str(e)}")
        return
    
    # Generate final synthetic images
    print("Generating synthetic images...")
    noise = tf.random.normal([NUM_SAMPLES_PER_CLASS * NUM_CLASSES, LATENT_DIM])
    generated_images = generator(noise, training=False).numpy()
    
    # Save generated images
    os.makedirs(os.path.join(OUTPUT_DIR, 'final'), exist_ok=True)
    for i, img in enumerate(generated_images):
        img = (img * 127.5 + 127.5).astype(np.uint8)  # Convert to 0-255 range
        img_pil = Image.fromarray(img)
        class_idx = i // NUM_SAMPLES_PER_CLASS
        class_name = class_names[class_idx]
        img_pil.save(os.path.join(OUTPUT_DIR, 'final', f'{class_name}_{i}.png'))
    
    print(f"Generated {len(generated_images)} images saved to {OUTPUT_DIR}/final")
    
    # Evaluation (using a subset of real images)
    print("Calculating evaluation metrics...")
    real_subset = images[:100]  # Use first 100 real images for comparison
    gen_subset = generated_images[:100]
    
    # Calculate SSIM
    ssim_score = calculate_ssim(real_subset, gen_subset)
    print(f"SSIM Score: {ssim_score:.4f}")
    
    # Calculate FID (may take some time)
    fid_score = calculate_fid(real_subset, gen_subset)
    print(f"FID Score: {fid_score:.2f}")
    
    # Save metrics to file
    with open(os.path.join(OUTPUT_DIR, 'metrics.txt'), 'w') as f:
        f.write(f"SSIM: {ssim_score:.4f}\n")
        f.write(f"FID: {fid_score:.2f}\n")
    
    print("All done!")

if __name__ == "__main__":
    main()
```

```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
import cv2
import time

# ================= CONFIGURATION =================
IMAGE_SIZE = 256                # Target resolution (256x256)
BATCH_SIZE = 16                 # Reduced for memory constraints
EPOCHS = 5                     # Minimum for high-res
LATENT_DIM = 256                # Larger noise vector
NUM_CLASSES = 7                 # From your dataset
STEPS_PER_EPOCH = 100           # Balance training time
CHECKPOINT_DIR = './training_checkpoints'
OUTPUT_DIR = './generated_images'

# ================= VERIFIED ARCHITECTURE =================
def make_generator_model():
    try:
        model = tf.keras.Sequential([
            layers.Dense(16*16*512, use_bias=False, input_shape=(LATENT_DIM,)),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            layers.Reshape((16, 16, 512)),
            
            layers.Conv2DTranspose(256, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(128, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(64, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(32, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(3, (5,5), strides=(1,1), padding='same', use_bias=False, activation='tanh')
        ])
        print("Generator architecture verified ✔")
        return model
    except Exception as e:
        print(f"Generator creation failed: {str(e)}")
        raise

def make_discriminator_model():
    try:
        model = tf.keras.Sequential([
            layers.Conv2D(64, (5,5), strides=(2,2), padding='same', input_shape=[IMAGE_SIZE, IMAGE_SIZE, 3]),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.Conv2D(128, (5,5), strides=(2,2), padding='same'),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.Conv2D(256, (5,5), strides=(2,2), padding='same'),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.Conv2D(512, (5,5), strides=(2,2), padding='same'),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.GlobalAveragePooling2D(),
            layers.Dense(1)
        ])
        print("Discriminator architecture verified ✔")
        return model
    except Exception as e:
        print(f"Discriminator creation failed: {str(e)}")
        raise

# ================= DATA PIPELINE =================
def smart_resize(img_path):
    try:
        img = Image.open(img_path).convert('RGB')
        img = ImageOps.fit(img, (IMAGE_SIZE, IMAGE_SIZE), method=Image.LANCZOS)
        img = (np.array(img)/127.5) - 1.0
        return img.astype(np.float32)
    except Exception as e:
        print(f"Error processing {img_path}: {str(e)}")
        return None

def load_dataset(data_dir):
    images = []
    print(f"Loading dataset from {data_dir}...")
    for class_name in os.listdir(data_dir):
        class_dir = os.path.join(data_dir, class_name)
        if not os.path.isdir(class_dir):
            continue
            
        for img_name in os.listdir(class_dir):
            img_path = os.path.join(class_dir, img_name)
            img = smart_resize(img_path)
            if img is not None:
                images.append(img)
    
    print(f"Loaded {len(images)} images after preprocessing")
    return np.array(images)

# ================= LOSS FUNCTIONS =================
def discriminator_loss(real_output, fake_output):
    real_loss = tf.keras.losses.BinaryCrossentropy(from_logits=True)(tf.ones_like(real_output), real_output)
    fake_loss = tf.keras.losses.BinaryCrossentropy(from_logits=True)(tf.zeros_like(fake_output), fake_output)
    return real_loss + fake_loss

def generator_loss(fake_output):
    return tf.keras.losses.BinaryCrossentropy(from_logits=True)(tf.ones_like(fake_output), fake_output)

# ================= TRAINING UTILITIES =================
def generate_samples(generator, epoch, num_samples=5):
    try:
        noise = tf.random.normal([num_samples, LATENT_DIM])
        generated_images = generator(noise, training=False)
        
        plt.figure(figsize=(10, 10))
        for i in range(generated_images.shape[0]):
            plt.subplot(1, num_samples, i+1)
            plt.imshow((generated_images[i] * 0.5 + 0.5))
            plt.axis('off')
        
        plt.savefig(os.path.join(OUTPUT_DIR, f'epoch_{epoch:04d}.png'))
        plt.close()
        print(f"Saved sample images for epoch {epoch}")
    except Exception as e:
        print(f"Error generating samples: {str(e)}")

def get_optimizers():
    gen_opt = tf.keras.optimizers.Adam(2e-4, beta_1=0.5)
    disc_opt = tf.keras.optimizers.Adam(2e-4, beta_1=0.5)
    return gen_opt, disc_opt

# ================= TRAINING LOOP =================
def train_step(images, generator, discriminator, gen_opt, disc_opt):
    noise = tf.random.normal([BATCH_SIZE, LATENT_DIM])
    
    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator(noise, training=True)
        
        real_output = discriminator(images, training=True)
        fake_output = discriminator(generated_images, training=True)
        
        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)
    
    gen_gradients = gen_tape.gradient(gen_loss, generator.trainable_variables)
    disc_gradients = disc_tape.gradient(disc_loss, discriminator.trainable_variables)
    
    gen_opt.apply_gradients(zip(gen_gradients, generator.trainable_variables))
    disc_opt.apply_gradients(zip(disc_gradients, discriminator.trainable_variables))
    
    return gen_loss, disc_loss

def train():
    try:
        # Initialize models
        generator = make_generator_model()
        discriminator = make_discriminator_model()
        gen_opt, disc_opt = get_optimizers()
        
        # Verify shapes
        test_noise = tf.random.normal([1, LATENT_DIM])
        test_output = generator(test_noise)
        print(f"Generator output shape: {test_output.shape}")
        
        # Load data
        dataset = load_dataset("/kaggle/input/mangoleaf-dataset/dataset")
        train_dataset = tf.data.Dataset.from_tensor_slices(dataset).shuffle(1000).batch(BATCH_SIZE)
        
        # Training
        for epoch in range(EPOCHS):
            start = time.time()
            total_gen_loss = 0
            total_disc_loss = 0
            
            for batch in train_dataset.take(STEPS_PER_EPOCH):
                gen_loss, disc_loss = train_step(batch, generator, discriminator, gen_opt, disc_opt)
                total_gen_loss += gen_loss
                total_disc_loss += disc_loss
            
            # Generate samples
            if epoch % 10 == 0:
                generate_samples(generator, epoch)
            
            print(f"Epoch {epoch+1}/{EPOCHS} | "
                  f"Gen Loss: {total_gen_loss/STEPS_PER_EPOCH:.4f} | "
                  f"Disc Loss: {total_disc_loss/STEPS_PER_EPOCH:.4f} | "
                  f"Time: {time.time()-start:.2f}s")
            
    except Exception as e:
        print(f"Training failed: {str(e)}")
        raise

```

```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
import cv2
import time

# ================= CONFIGURATION =================
IMAGE_SIZE = 256                # Target resolution (256x256)
BATCH_SIZE = 16                 # Reduced for memory constraints
EPOCHS = 5                    # Minimum for high-res
LATENT_DIM = 256                # Larger noise vector
NUM_CLASSES = 7                 # From your dataset
STEPS_PER_EPOCH = 100           # Balance training time
CHECKPOINT_DIR = './training_checkpoints'
OUTPUT_DIR = './generated_images'

# ================= VERIFIED ARCHITECTURE =================
def make_generator_model():
    try:
        model = tf.keras.Sequential([
            layers.Dense(16*16*512, use_bias=False, input_shape=(LATENT_DIM,)),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            layers.Reshape((16, 16, 512)),
            
            layers.Conv2DTranspose(256, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(128, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(64, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(32, (5,5), strides=(2,2), padding='same', use_bias=False),
            layers.BatchNormalization(),
            layers.LeakyReLU(0.2),
            
            layers.Conv2DTranspose(3, (5,5), strides=(1,1), padding='same', use_bias=False, activation='tanh')
        ])
        print("Generator architecture verified ✔")
        return model
    except Exception as e:
        print(f"Generator creation failed: {str(e)}")
        raise

def make_discriminator_model():
    try:
        model = tf.keras.Sequential([
            layers.Conv2D(64, (5,5), strides=(2,2), padding='same', input_shape=[IMAGE_SIZE, IMAGE_SIZE, 3]),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.Conv2D(128, (5,5), strides=(2,2), padding='same'),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.Conv2D(256, (5,5), strides=(2,2), padding='same'),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.Conv2D(512, (5,5), strides=(2,2), padding='same'),
            layers.LeakyReLU(0.2),
            layers.Dropout(0.3),
            
            layers.GlobalAveragePooling2D(),
            layers.Dense(1)
        ])
        print("Discriminator architecture verified ✔")
        return model
    except Exception as e:
        print(f"Discriminator creation failed: {str(e)}")
        raise

# ================= DATA PIPELINE =================
def smart_resize(img_path):
    try:
        img = Image.open(img_path).convert('RGB')
        img = ImageOps.fit(img, (IMAGE_SIZE, IMAGE_SIZE), method=Image.LANCZOS)
        img = (np.array(img)/127.5) - 1.0
        return img.astype(np.float32)
    except Exception as e:
        print(f"Error processing {img_path}: {str(e)}")
        return None

def load_dataset(data_dir):
    images = []
    print(f"Loading dataset from {data_dir}...")
    for class_name in os.listdir(data_dir):
        class_dir = os.path.join(data_dir, class_name)
        if not os.path.isdir(class_dir):
            continue
            
        for img_name in os.listdir(class_dir):
            img_path = os.path.join(class_dir, img_name)
            img = smart_resize(img_path)
            if img is not None:
                images.append(img)
    
    print(f"Loaded {len(images)} images after preprocessing")
    return np.array(images)

# ================= LOSS FUNCTIONS =================
def discriminator_loss(real_output, fake_output):
    real_loss = tf.keras.losses.BinaryCrossentropy(from_logits=True)(tf.ones_like(real_output), real_output)
    fake_loss = tf.keras.losses.BinaryCrossentropy(from_logits=True)(tf.zeros_like(fake_output), fake_output)
    return real_loss + fake_loss

def generator_loss(fake_output):
    return tf.keras.losses.BinaryCrossentropy(from_logits=True)(tf.ones_like(fake_output), fake_output)

# ================= TRAINING UTILITIES =================
def generate_samples(generator, epoch, num_samples=5):
    try:
        noise = tf.random.normal([num_samples, LATENT_DIM])
        generated_images = generator(noise, training=False)
        
        plt.figure(figsize=(10, 10))
        for i in range(generated_images.shape[0]):
            plt.subplot(1, num_samples, i+1)
            plt.imshow((generated_images[i] * 0.5 + 0.5))
            plt.axis('off')
        
        plt.savefig(os.path.join(OUTPUT_DIR, f'epoch_{epoch:04d}.png'))
        plt.close()
        print(f"Saved sample images for epoch {epoch}")
    except Exception as e:
        print(f"Error generating samples: {str(e)}")

def get_optimizers():
    gen_opt = tf.keras.optimizers.Adam(2e-4, beta_1=0.5)
    disc_opt = tf.keras.optimizers.Adam(1e-5, beta_1=0.5)
    return gen_opt, disc_opt

# ================= TRAINING LOOP =================
def train_step(images, generator, discriminator, gen_opt, disc_opt):
    noise = tf.random.normal([BATCH_SIZE, LATENT_DIM])
    
    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator(noise, training=True)
        
        real_output = discriminator(images, training=True)
        fake_output = discriminator(generated_images, training=True)
        
        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)
    
    gen_gradients = gen_tape.gradient(gen_loss, generator.trainable_variables)
    disc_gradients = disc_tape.gradient(disc_loss, discriminator.trainable_variables)
    
    gen_opt.apply_gradients(zip(gen_gradients, generator.trainable_variables))
    disc_opt.apply_gradients(zip(disc_gradients, discriminator.trainable_variables))
    
    return gen_loss, disc_loss
def train():
    try:
        # Initialize models
        generator = make_generator_model()
        discriminator = make_discriminator_model()
        gen_opt, disc_opt = get_optimizers()
        
        # Verify shapes
        test_noise = tf.random.normal([1, LATENT_DIM])
        test_output = generator(test_noise)
        print(f"Generator output shape: {test_output.shape}")
        
        # Load data
        dataset = load_dataset("/kaggle/input/mango-leaf/Train")
        train_dataset = tf.data.Dataset.from_tensor_slices(dataset).shuffle(1000).batch(BATCH_SIZE)
        
        # Training
        for epoch in range(EPOCHS):
            start = time.time()
            total_gen_loss = 0
            total_disc_loss = 0
            
            for batch in train_dataset.take(STEPS_PER_EPOCH):
                gen_loss, disc_loss = train_step(batch, generator, discriminator, gen_opt, disc_opt)
                total_gen_loss += gen_loss
                total_disc_loss += disc_loss
            
            # Display generated images every 2 epochs with class labels
            if (epoch + 1) % 2 == 0:
                plt.figure(figsize=(15, 15))
                num_samples = min(9, NUM_SAMPLES_PER_CLASS)  # Display up to 9 samples
                test_noise = tf.random.normal([num_samples, LATENT_DIM])
                generated_images = generator(test_noise, training=False)
                
                for i in range(num_samples):
                    plt.subplot(3, 3, i+1)  # 3x3 grid
                    plt.imshow((generated_images[i] * 0.5 + 0.5))
                    
                    # Add sample number and epoch info
                    plt.title(f"Epoch {epoch+1}\nSample {i+1}", fontsize=10)
                    plt.axis('off')
                
                plt.suptitle(f"Generated Samples (Epoch {epoch+1})", fontsize=16, y=1.02)
                plt.tight_layout()
                plt.show()
            
            print(f"Epoch {epoch+1}/{EPOCHS} | "
                  f"Gen Loss: {total_gen_loss/STEPS_PER_EPOCH:.4f} | "
                  f"Disc Loss: {total_disc_loss/STEPS_PER_EPOCH:.4f} | "
                  f"Time: {time.time()-start:.2f}s")
            
    except Exception as e:
        print(f"Training failed: {str(e)}")
        raise

```

```python
# ================= MAIN EXECUTION =================
if __name__ == "__main__":
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    try:
        print("Starting training...")
        train()
    except Exception as e:
        print(f"Fatal error: {str(e)}")
```
